Importing the libraries


```python
import numpy as np # numpy is used to create numpy arrays
import pandas as pd # pandas is used to create data frames {structured tables}
import matplotlib as plt #
import seaborn as sns
from sklearn.model_selection import train_test_split # The function train_test_split is used to split data into training data and testing data
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score # This accuracy score is used to evaluate our model like how many good or correct predictions in model
```

DATA COLLECTION AND PROCESSING PART

```python
# Load the data of the csv file to the Pandas DataFrame
titanic_data = pd.read_csv("/content/train.csv")
```

```python
# The starting 5 rows of the dataframe
titanic_data.head()
```

```python
# how many rows and how many columns ?
titanic_data.shape
```

```python
# Extract some information about the data
titanic_data.info()
```

```python
# Check the number of missing values in each column
titanic_data.isnull().sum()
```

  HANDLING THE MISSING VALUES

```python
#Drop the Cabin column
titanic_data = titanic_data.drop(columns='Cabin', axis=1)
```

```python
#Replacing the missing value in "Age" column with mean value
titanic_data['Age'].fillna(titanic_data['Age'].mean(), inplace=True)
```

```python
#Finding the mode value of "Embarked" column
print(titanic_data['Embarked'].mode())
```

```python
print(titanic_data['Embarked'].mode()[0])
```

```python
#Replacing the missing values in 'Embarked' column with mode value
titanic_data['Embarked'].fillna(titanic_data['Embarked'].mode()[0],inplace =True)
```

```python
# check the number of missing value in each column
titanic_data.isnull().sum()
```

DATA ANALYSIS

```python
#getting some statistical measures about the data
titanic_data.describe()
```

```python
#finding the number of people survived and not survived
titanic_data['Survived'].value_counts()
```

ENCODING THE CATEGORICAL COLUMNS

```python
titanic_data['Sex'].value_counts()
```

```python
titanic_data['Embarked'].value_counts()
```

```python
# Converting categorical columns
titanic_data.replace({'Sex':{'male':0,'female':1}, 'Embarked':{'S':0,'C':1,'Q':2}}, inplace=True)
```

```python
titanic_data.head()
```

SEPARATING FEATURES & TARGET

```python
X = titanic_data.drop(columns=['PassengerId','Name','Ticket','Survived'], axis=1)
Y = titanic_data['Survived']
```

```python
print(X)
```

```python
print(Y)
```

SPLITTING THE DATA INTO TRAINING DATA & TEST DATA

```python
X_train , X_test , Y_train , Y_test = train_test_split(X,Y, test_size=0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

MODEL TRAINING

LOGISTIC REGRESSION

```python
model = LogisticRegression()
```

```python
# training the logistic Regression model with trainig data
model.fit(X_train, Y_train)
```

MODEL EVALUATION

ACCURACY SCORE

```python
# accuracy on training data
X_train_prediction = model.predict(X_train)
```

```python
print(X_train_prediction)
```

```python
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print('Accuracy score of training data : ', training_data_accuracy)
```

```python
# accuracy score on test data
X_test_prediction = model.predict(X_test)
```

```python
print(X_test_prediction)
```

```python
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print('Accuracy score of test data : ', test_data_accuracy)
```
